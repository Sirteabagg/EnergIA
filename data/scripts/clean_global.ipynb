```python
import pandas as pd
import numpy as np
```

```python
# Load data
df = pd.read_csv('../raw/energy_data_global.csv')
```

```python
df.head(10)
```

```python
df.describe()
```

```python
df = df.dropna()
df.describe()
```

```python
# Suppression des consommations négatives
df = df[df['power_consumption'] >= 0]
df.describe()
```

```python
# Suppression des consommations négatives
df = df[df['temperature'] >= 0]
df.describe()
```

```python
# ecart interquantile pour determiner un seuil max
Q1 = df['power_consumption'].quantile(0.25)
Q3 = df['power_consumption'].quantile(0.75)
IQR = Q3 - Q1
upper_limit = Q3 + 1.5 * IQR
df = df[df['power_consumption'] < upper_limit]
df.describe()
```

```python
# Sauvegarde des données
df.to_csv('../processed/clean_global.csv', index=False)
```
