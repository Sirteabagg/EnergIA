```python
import pandas as pd
```

### Charger les données

```python
df = pd.read_csv('../processed/clean_global.csv')
df.info()
```

### Indicateur global

```python
df.describe()
```

### Conversion en datetime puis ajout d'une colonne 'date'

```python
df['date'] = pd.to_datetime(df['timestamp']).dt.date
```

### Calcul des indicateurs pour le power_consumption

```python

power_consumption = (
    df.groupby(['building_id', 'date'])
      .agg(
        moyenne=('power_consumption', 'mean'),
        minimum=('power_consumption', 'min'),
        maximum=('power_consumption', 'max'),
        ecart_type=('power_consumption', 'std')
      )
      .reset_index()
)
power_consumption
```

### Calcul des indicateurs pour l'humidité

```python
humidity = (
    df.groupby(['building_id', 'date'])
      .agg(
        moyenne=('humidity', 'mean'),
        minimum=('humidity', 'min'),
        maximum=('humidity', 'max'),
        ecart_type=('humidity', 'std')
      )
      .reset_index()
)
humidity
```

### Calcul des indicateurs pour la température

```python
temperature = (
    df.groupby(['building_id', 'date'])
      .agg(
        moyenne=('temperature', 'mean'),
        minimum=('temperature', 'min'),
        maximum=('temperature', 'max'),
        ecart_type=('temperature', 'std')
      )
      .reset_index()
)
temperature
```
