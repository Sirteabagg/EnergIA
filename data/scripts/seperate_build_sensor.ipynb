```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('../raw/energy_data_global.csv')
```

```python
# seperate unique building name and give id

result = df["building_id"].unique()
result = np.sort(result)

df_building = pd.DataFrame(zip(range(len(result)), result), columns=["building_id", "name"])
```

```python
# seperate unique building name and give id


result = df["sensor_id"].unique()
result = np.sort(result)

df_sensor = pd.DataFrame(zip(range(len(result)), result), columns=["sensor_id", "name"])
```

```python
df_building.to_csv('../processed/building.csv', index=False)
df_sensor.to_csv('../processed/sensor.csv', index=False)
```
